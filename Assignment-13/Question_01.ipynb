```python
# 1) Replace Nan with 0 and Interchangea 3 rows and 3 columns of 2D array [[6, -8, 73, -110], [np.nan, -8, 0, 94]]
```

```python
import numpy as np
```

```python

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

print("Original Array:")
print(arr)

res = np.nan_to_num(arr)
print("\nAfter Replacing NaN with 0:")
print(res)

print("\nAfter Interchanging Rows and Columns:")
print(np.transpose(res))
```
