```python
import numpy as np
import matplotlib.pyplot as plt

subjects = ["Maths", "Physics", "Chemistry", "C Programming", "English"]

sem1 = [78, 82, 75, 88, 80]
sem2 = [85, 79, 81, 92, 84]

x = np.arange(len(subjects))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(x - width/2, sem1, width, label="Semester 1")
plt.bar(x + width/2, sem2, width, label="Semester 2")

plt.title("Semester 1 vs Semester 2 Results")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(x, subjects, rotation=20)
plt.legend()
plt.grid(True)

plt.show()
```
