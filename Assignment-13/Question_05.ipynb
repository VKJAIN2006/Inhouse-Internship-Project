```python
import numpy as np
from scipy.stats import mode
```

```python
arr1 = np.array([3, 4]) 
arr2 = np.array([1, 0]) 
print(arr1)
print(arr2)

avg = (arr1 + arr2) / 2 
print("Average of NumPy arrays:\n", avg)
```

```python
a = np.array([[1,2,3],[10,20,30]])
b = np.array([[40,50,60],[70,80,90]])

# Average of two arrays
avg = (a + b) / 2
print("Average:\n", avg)

# Mean
print("Mean of a:", np.mean(a))
print("Mean of b:", np.mean(b))

# Median
print("Median of a:", np.median(a))
print("Median of b:", np.median(b))

# Mode
print("Mode of a:", mode(a, axis=None))
print("Mode of b:", mode(b, axis=None))
```
