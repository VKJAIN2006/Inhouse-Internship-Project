```python
#  4) Write a Python program to calculate the total, highest value, and lowest value from a tuple of integers 
# without using the built-in sum(), max(), and min() functions.
```

```python
numbers = (12, 45, 7, 89, 23, 56)

total = 0
highest = numbers[0]
lowest = numbers[0]

# Calculate total, highest, and lowest
for num in numbers:
    total += num

    if num > highest:
        highest = num

    if num < lowest:
        lowest = num

# Display results
print("Tuple:", numbers)
print("Total =", total)
print("Highest Value =", highest)
print("Lowest Value =", lowest)
```
