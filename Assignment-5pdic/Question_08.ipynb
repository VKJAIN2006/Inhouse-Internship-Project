```python
# 8) Write a Python program to find all elements that are common to both sets. 
# s1 = {1, 2, 3, 4} s2 = {3, 4, 5, 6} 9) Write a Python program to find all elements that are in eith

```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

# Find common elements in both sets
common_elements = s1.intersection(s2)

# Display the result
print("Common elements:", common_elements)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

# Find all elements that are in either set
all_elements = s1.union(s2)

# Display the result
print("Union of sets:", all_elements)
```
